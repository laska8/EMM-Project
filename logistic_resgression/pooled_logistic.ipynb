```python
from pathlib import Path

import numpy as np
import pandas as pd
import statsmodels.formula.api as smf

DATA = Path(r"C:\Users\tahai\Desktop\EMM-Project\eda_outputs\reciprocity_return.csv")

assert DATA.exists(), f"File not found: {DATA}"
```

```python
subreddit_sizes = pd.Series(dtype="float64")

for chunk in pd.read_csv(
    DATA,
    usecols=["subreddit"],
    chunksize=200_000,
    keep_default_na=False,
):
    counts = chunk["subreddit"].value_counts()

    subreddit_sizes = subreddit_sizes.add(
        counts,
        fill_value=0
    )

    selected_subreddits = subreddit_sizes.index
    print("Subreddits:", len(selected_subreddits))
    display(subreddit_sizes.loc[selected_subreddits])
```

```python
print("Output test")
print(subreddit_sizes.nlargest(20).to_string())
```

```python
columns = [
    "user",
    "subreddit",
    "month",
    "replies_sent",
    "replies_received",
    "reciprocal_partners",
    "returned_next_month",
]

target_rows = 50_000

sample_fraction = 1.0

parts = []

for chunk in pd.read_csv(
    DATA,
    usecols=columns,
    dtype={"user": str, "subreddit": str},
    chunksize=200_000,
    keep_default_na=False,
):
    chunk = chunk[
        chunk["subreddit"].isin(selected_subreddits)
    ].copy()

    user_hash = pd.util.hash_pandas_object(
        chunk["user"],
        index=False,
    )

    keep_user = (
        user_hash % 1_000_003
    ) < sample_fraction * 1_000_003

    parts.append(chunk.loc[keep_user])

df = pd.concat(parts, ignore_index=True)
del parts

df["user"] = df["user"].astype("category")
df["subreddit"] = df["subreddit"].astype("category")

print("Rows:", len(df))
print("Users:", df["user"].nunique())
print("Subreddits:", df["subreddit"].nunique())

display(df.head())
```

```python
df["recip_binary"] = (
    df["reciprocal_partners"] > 0
).astype(int)

df["log_sent"] = np.log1p(df["replies_sent"])
df["log_received"] = np.log1p(df["replies_received"])

assert not df.empty
assert not df.isna().any().any()

assert not df.duplicated(
    ["user", "subreddit", "month"]
).any(), "Duplicate user-subreddit-month rows found."

assert set(df["returned_next_month"].unique()) == {0, 1}
assert set(df["recip_binary"].unique()) == {0, 1}

assert (
    df[["replies_sent", "replies_received", "reciprocal_partners"]]
    >= 0
).all().all()

assert np.isfinite(
    df[["log_sent", "log_received"]].to_numpy()
).all()

for column in ["subreddit", "month"]:
    variation = df.groupby(column)["returned_next_month"].nunique()
    assert variation.eq(2).all(), (
        f"Some {column} groups contain only one return outcome."
    )

display(
    pd.crosstab(
        df["recip_binary"],
        df["returned_next_month"],
    )
)
```

```python
print("Kernel is responding", flush=True)
print("Data shape:", df.shape, flush=True)
```

```python
import pyfixest as pf

model = pf.feglm(
    "returned_next_month ~ recip_binary + log_sent + log_received"
    " | month + subreddit",
    data=df,
    family="logit",
    vcov={"CRV1": "user"},
    copy_data=False,
    fixef_rm="none",
    iwls_maxiter=100,
)

assert model.convergence, "Model did not converge."

display(model.tidy())
```

```python
from scipy.special import expit

eta = model.predict(type="link")

assert len(eta) == len(df), "Some observations were excluded."
assert np.isfinite(eta).all(), "Some fitted values are missing or invalid."

beta = model.coef().loc["recip_binary"]

score_without_reciprocity = (
    eta - beta * df["recip_binary"].to_numpy()
)

average_difference = (
    expit(score_without_reciprocity + beta)
    - expit(score_without_reciprocity)
).mean()

print(
    "Adjusted average difference:",
    f"{100 * average_difference:.2f} percentage points",
)
```
